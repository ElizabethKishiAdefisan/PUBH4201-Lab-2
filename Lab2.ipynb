## Setup
#### Before this Lab is started, I am going to load the data

```python
import pandas as pd
import matplotlib.pyplot as plt
pbc = pd.read_csv("pbc1.csv")
```

##### This biostatics data set shows the survival and status of cirhossis patient per mayo clinic

### Nontrivial analyis
##### Here I am going to do some non-trivial analysis. In this lab, I choose to focus on ascites counts

```python

ascites_counts = pbc.groupby('ascites').size().round().astype('Int64')
print(ascites_counts)
```

##### This is the counts/status of ascites. Because this data is numeric, It doesn't really tell us much. I want to know what "0.0" and "1.0" means and how that relates to cirhossis. Acording to the data set, Ascites are grouped by status, 0 meaning no ascites have developed and 1 meaning some have. 

##### Now that we know that infomration, lets check out the percentage of those who have and have not developed ascites

```python
status_of_ascites = ascites_counts.rename(
    { 
        0: "Not developed ascites",
        1: "Has developed ascites"
    }
)
plt.figure(figsize=(7, 7))
plt.pie(status_of_ascites, labels=status_of_ascites.index, autopct="%1.1f%%")
plt.title("Status of Ascites")
plt.show()    
```

### What about per stage
##### In this data set, we have also deocumented the 4 stages of liver disease/failure. Because we know that ascites presents can be a hallmark for liver disease, we want to see the percentage of those with/without ascites and what stage of liver disease they are in

```python
status_per_stage = pbc.groupby('stage')['ascites'].value_counts()
renamed_status = status_per_stage.rename(
    { 
        0: "Not developed ascites",
        1: "Has developed ascites",
    }
)
plt.pie(renamed_status, labels=renamed_status.index,autopct="%1.1f%%")
plt.title("Status of Ascites per stage of liver failure")
plt.show() 
```
